```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.metrics import accuracy_score
```

```python
wine = load_wine()

df = pd.DataFrame(wine.data, columns=wine.feature_names)

df['target'] = wine.target

print("Primeras filas del dataset:")
print(df.head())
```

```python
primera_columna = df.columns[0]

df.loc[0:4, primera_columna] = np.nan

print("\nCantidad de NaN antes de imputar:")
print(df.isnull().sum())
```

```python
media = df[primera_columna].mean()

df[primera_columna] = df[primera_columna].fillna(media)

print("\nValores nulos después de la imputación:")
print(df.isnull().sum())
```

```python
X = df[wine.feature_names]
y = df['target']
scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled,
    y,
    test_size=0.25,
    random_state=42
)
modelo = DecisionTreeClassifier(random_state=42)

modelo.fit(X_train, y_train)
y_pred = modelo.predict(X_test)

accuracy = accuracy_score(y_test, y_pred)

print("\nExactitud del modelo:")
print(f"Accuracy = {accuracy:.4f}")
```

```python
plt.figure(figsize=(20,10))

plot_tree(
    modelo,
    feature_names=wine.feature_names,
    class_names=wine.target_names,
    filled=True
)

plt.title("Árbol de Decisión - Dataset Wine")
plt.show()
```

```python
raiz = wine.feature_names[modelo.tree_.feature[0]]

print("\nCaracterística utilizada en el nodo raíz:")
print(raiz)
```
